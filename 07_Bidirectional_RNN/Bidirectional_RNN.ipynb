```python
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing import sequence
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding,SimpleRNN Bidirectional,Dense,LSTM,GRU
```

```python
(x_train,y_train),(x_test,y_test)=imdb.load_data(num_words=10000)
x_train=sequence.pad_sequences(x_train,maxlen=100,padding='post',truncating='post')
x_test=sequence.pad_sequences(x_test,maxlen=100,padding='post',truncating='post')
```

```python
#BIDIRECTIONAL
model=Sequential([
    Embedding(10000,input_length=100,output_dim=32), #Embedding layer to convert words to vectors
    SimpleRNN(5), #RNN layer with 5 units
    Dense(1,activation='sigmoid') #Dense layer with 1 unit
])
```

```python
model.summary()
```

```python
#BIDIRECTIONAL SIMPLE RNN
model=Sequential([
    Embedding(10000,input_length=100,output_dim=32), #Embedding layer to convert words to vectors
    Bidirectional(SimpleRNN(5)), #RNN layer with 5 units
    Dense(1,activation='sigmoid') #Dense layer with 1 unit
])
```

```python
#LSTM
model=Sequential([
    Embedding(10000,input_length=100,output_dim=32), #Embedding layer to convert words to vectors
    Bidirectional(LSTM(5)), #RNN layer with 5 units
    Dense(1,activation='sigmoid') #Dense layer with 1 unit
])
```

```python
#GRU
model=Sequential([
    Embedding(10000,input_length=100,output_dim=32), #Embedding layer to convert words to vectors
    Bidirectional(GRU(5)), #RNN layer with 5 units
    Dense(1,activation='sigmoid') #Dense layer with 1 unit
])
```

```python
model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
history=model.fit(x_train,y_train,epochs=5,batch_size=32,validation_split=0.2)
```
