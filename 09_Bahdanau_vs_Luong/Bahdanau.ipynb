```python
# Bahdanau (Additive) Attention
This notebook implements Bahdanau Attention from scratch using TensorFlow Subclassing.
It uses a Feed-Forward Neural Network to calculate the attention scores.
```

```python
# %%
import tensorflow as tf

class BahdanauAttention(tf.keras.layers.Layer):
    def __init__(self, units):
        super(BahdanauAttention, self).__init__()
        # W1 for the Decoder Hidden State
        self.W1 = tf.keras.layers.Dense(units)
        # W2 for the Encoder Hidden States
        self.W2 = tf.keras.layers.Dense(units)
        # V for the final score output
        self.V = tf.keras.layers.Dense(1)

    def call(self, query, values):
        # Expand query to broadcast across time steps: (Batch, 1, Hidden Size)
        query_with_time_axis = tf.expand_dims(query, 1)

        # 1. Calculate Score: V * tanh(W1(query) + W2(values))
        score = self.V(tf.nn.tanh(self.W1(query_with_time_axis) + self.W2(values)))

        # 2. Calculate Attention Weights (Softmax)
        attention_weights = tf.nn.softmax(score, axis=1)

        # 3. Create Context Vector (Multiply weights by values and sum)
        context_vector = attention_weights * values
        context_vector = tf.reduce_sum(context_vector, axis=1)

        return context_vector, attention_weights
```

```python
# %%
# --- TEST THE BAHDANAU LAYER ---
print("🧪 Testing Bahdanau Attention")

# Dummy data: Batch=1, Time Steps=5, Hidden Size=10
sample_query = tf.random.normal((1, 10))
sample_values = tf.random.normal((1, 5, 10))

attention_layer = BahdanauAttention(units=10)
context, weights = attention_layer(sample_query, sample_values)

print(f"Query Shape: {sample_query.shape}")
print(f"Values (Encoder States) Shape: {sample_values.shape}")
print(f"Context Vector Shape: {context.shape}")
print(f"Attention Weights Shape: {weights.shape}")
```
