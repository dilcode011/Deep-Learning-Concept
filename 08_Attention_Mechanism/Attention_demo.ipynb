```python
import numpy as np

# Set print options for cleaner output
np.set_printoptions(precision=3, suppress=True)

# Let's imagine a tiny sentence with 3 words: "Bank of River"
# We will represent each word as a simple 4-dimensional embedding vector.
word_1 = np.array([1, 0, 1, 0]) # "Bank"
word_2 = np.array([0, 1, 0, 1]) # "of"
word_3 = np.array([1, 1, 1, 1]) # "River"

# Combine into a sequence (Batch Size = 1, Seq Length = 3, Emb Size = 4)
inputs = np.array([word_1, word_2, word_3])
print("Input Sequence Shape:", inputs.shape)
print("Input Sequence:\n", inputs)
```

```python
# In a real neural network, Q, K, and V are created by multiplying the inputs 
# by weight matrices that the model learns during training.
# For this concept demo, we will generate them randomly to represent the 
# learned state of the network.

np.random.seed(42)

# Create random weight matrices for Q, K, V
W_q = np.random.randint(1, 4, (4, 3)) # Maps 4D input to 3D Query
W_k = np.random.randint(1, 4, (4, 3)) # Maps 4D input to 3D Key
W_v = np.random.randint(1, 4, (4, 3)) # Maps 4D input to 3D Value

# Calculate Q, K, V by multiplying inputs by weights
Q = np.dot(inputs, W_q)
K = np.dot(inputs, W_k)
V = np.dot(inputs, W_v)

print("Queries (Q):\n", Q)
print("\nKeys (K):\n", K)
print("\nValues (V):\n", V)
```

```python
# STEP 1: Dot Product of Q and K-transpose
# This calculates how much every word "attends" to every other word
scores = np.dot(Q, K.T)
print("Raw Attention Scores (Q * K^T):\n", scores)

# STEP 2: Scale the scores
# We divide by the square root of the dimension of the Keys (which is 3 here)
d_k = K.shape[1]
scaled_scores = scores / np.sqrt(d_k)

# STEP 3: Softmax
# Convert scores to percentages (probabilities that sum to 1)
def softmax(x):
    e_x = np.exp(x - np.max(x, axis=-1, keepdims=True))
    return e_x / e_x.sum(axis=-1, keepdims=True)

attention_weights = softmax(scaled_scores)
print("\nAttention Weights (Softmax applied):\n", attention_weights)
print("(Notice how each row sums to 1.0)")

# STEP 4: Multiply by Values
# This creates the final output context vector for each word
output = np.dot(attention_weights, V)
print("\nFinal Attention Output:\n", output)
```

```python
# In real projects, you don't write that math manually. 
# Keras has a built-in layer for it!

import tensorflow as tf
from tensorflow.keras.layers import Attention

# Create dummy tensors (Batch=1, SeqLen=3, Dim=4)
query_tensor = tf.random.normal([1, 3, 4])
value_tensor = tf.random.normal([1, 3, 4])
# (If key is not provided, it defaults to using the value tensor as the key)

# Initialize the Attention layer
# use_scale=True implements the division by sqrt(d_k) we did above
attention_layer = Attention(use_scale=True)

# Pass the tensors through the layer
context_vector = attention_layer([query_tensor, value_tensor])

print("Keras Attention Output Shape:", context_vector.shape)
```
