```python
from tensorflow.keras.layers import GRU
```

```python
# 1. Create Dummy Input (1 Sample, 5 TimeSteps, 10 Features)
dummy_input = np.random.random((1, 5, 10)).astype(np.float32)

inputs = Input(shape=(5, 10))
```

```python
# 2. Define GRU Layer
# We use return_state=True to see what it gives us back
gru_out, gru_state = GRU(units=3, return_state=True)(inputs)

model_gru = Model(inputs=inputs, outputs=[gru_out, gru_state])
```

```python
# 3. Run it
output, state = model_gru.predict(dummy_input)

print("--- GRU Results ---")
print(f"Output Shape: {output.shape}")
print(f"State Shape:  {state.shape}")

print("\n--- Verification ---")
print(f"Output Vector: \n{output[0]}")
print(f"State Vector:  \n{state[0]}")
print(f"\nAre they identical? {np.allclose(output, state)}")
# Expected: True (because GRU merges Cell and Hidden states)
```
